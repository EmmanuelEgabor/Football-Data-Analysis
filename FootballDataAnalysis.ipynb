## feature engineering

### creating age group column


```python
# defining a function called age group
def age_group(n):
    if n <= 17:
        return "under 17"
    elif n <= 21:
        return "under 21"
    else:
        return "senior"
```

```python
#applying the function to the df to create a new column called age_group
football_df["Age_Group"] = football_df["Age"].apply(age_group)
```

```python
# player age table
player_age_table = football_df[["Name", "Age", "Age_Group"]]
```

```python
player_age_table
```

### Player category


```python
# defining a function called player category
def player_cat(n):
    if n >= 80:
        return "Elite"
    elif n >= 70:
        return "Excellent"
    elif n >= 50:
        return "Average"
    else:
        return "Below Average"
```

```python
#applying the function to the df to create a new column called player category
football_df["player_category"] = football_df["Overall"].apply(player_cat)
```

## Data Analysis

## Nationality that produces the highest rated players


```python
nationality_rating = football_df.groupby("Nationality")["Overall"].mean()
```

```python
nationality_rating = (
    football_df.groupby("Nationality")["Overall"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
nationality_rating
```

```python
# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(nationality_rating.index, nationality_rating.values)

# Add titles and labels
plt.title("Top 10 Nationalities by Average Overall Rating", fontsize=14)
plt.xlabel("Nationality", fontsize=12)
plt.ylabel("Average Overall Rating", fontsize=12)

# Rotate x-axis labels for readability
plt.xticks(rotation=45, ha="right")

# Add value labels on top of bars
for i, value in enumerate(nationality_rating.values):
    plt.text(i, value + 0.1, f"{value:.1f}", ha="center", fontsize=10)

plt.tight_layout()
plt.show()
```

## Players position with the highest overall

```python
football_df[["Preferred Positions", "Preferred Positions Type"]]
```

```python
preferred_positions_type = football_df.groupby("Preferred Positions Type")["Overall"].mean()
```

```python
preffered_position_rating = (
    football_df.groupby("Preferred Positions Type")["Overall"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
preffered_position_rating
```

```python
# Plot pie chart in Percentages
plt.figure(figsize=(8, 8))

plt.pie(
    preffered_position_rating.values,
    labels=preffered_position_rating.index,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.8
)

plt.title("Top 10 Preferred Position Types by Average Overall Rating")
plt.axis('equal')  # Makes the pie chart circular

plt.show()
```

```python
# Function to display actual values instead of percentages
def show_value(pct):
    total = sum(preffered_position_rating.values)
    value = pct * total / 100
    return f"{value:.2f}"

plt.figure(figsize=(8, 8))

plt.pie(
    preffered_position_rating.values,
    labels=preffered_position_rating.index,
    autopct=show_value,
    startangle=90,
    shadow=True
)

plt.title("Top 10 Preferred Position Types by Average Overall Rating")
plt.axis("equal")

plt.show()
```

```python
# Pie chart showing Age_Group with highets performance
# Function to display actual values
def show_value(pct):
    total = age_group.sum()
    value = pct * total / 100
    return f"{value:.2f}"

plt.figure(figsize=(7, 7))

plt.pie(
    age_group.values,
    labels=age_group.index,
    autopct=show_value,
    startangle=90,
    shadow=True,
    explode=[0.03, 0.03, 0.03]
)

plt.title("Average Overall Rating by Age Group")
plt.axis("equal")

plt.show()
```

```python
# Age group average ratings
age_group = (
    football_df.groupby("Age_Group")["Overall"]
    .mean()
)

plt.figure(figsize=(7, 5))

bars = plt.bar(
    age_group.index,
    age_group.values,
    width=0.6
)

plt.title("Average Overall Rating by Age Group", fontsize=14)
plt.xlabel("Age Group", fontsize=12)
plt.ylabel("Average Overall Rating", fontsize=12)

# Display values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.3,
        f"{height:.2f}",
        ha="center",
        fontsize=10
    )

plt.ylim(0, max(age_group.values) + 10)
plt.tight_layout()
plt.show()
```

```python
# Top 10 nationalities by number of players
nationality = (
    football_df.groupby("Nationality")["Overall"]
    .count()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    nationality.index,
    nationality.values,
    width=0.7
)

plt.title("Top 10 Nationalities by Number of Players", fontsize=14)
plt.xlabel("Nationality", fontsize=12)
plt.ylabel("Number of Players", fontsize=12)

plt.xticks(rotation=45, ha='right')

# Add value labels
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 2,
        f"{int(height)}",
        ha='center',
        fontsize=10
    )

plt.tight_layout()
plt.show()
```
